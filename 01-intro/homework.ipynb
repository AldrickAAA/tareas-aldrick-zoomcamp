## Q3. Fuel types

```python
fuel_types = df["fuel_type"].nunique()
print("Tipos de combustible:", df["fuel_type"].unique())
```

## Q4. Missing values

```python
missing_cols = df.isna().any()
count_missing_cols = missing_cols.sum()
print("Missing values:\n", df.isna().sum()[missing_cols])
print(count_missing_cols)
```

## Q5. Max fuel efficiency

```python
max_mpg_asia = df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max()
print(max_mpg_asia)
```

## Q6. Median value of horsepower

```python
median_before = df["horsepower"].median()
mode_hp = df["horsepower"].mode().iloc[0]
df_filled = df["horsepower"].fillna(mode_hp)
median_after = df_filled.median()
print(f"Mediana antes: {median_before}")
print(f"Moda usada:   {mode_hp}")
print(f"Mediana despues: {median_after}")
```

## Q7. Sum of weights

```python
subset_asia = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7)
X = subset_asia.to_numpy()
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
sum_weights = w.sum()
print("Vector de pesos w:", w)
print(f"Suma de pesos: {sum_weights:.4f}")
```

## Marco de validación: Split Train / Validation / Test

```python
def split_dataset(df, seed):
    # Split 60/20/20 exactamente como en la lección
    n = len(df)
    n_val = int(n * 0.2)
    n_test = int(n * 0.2)
    n_train = n - n_val - n_test

    np.random.seed(seed)
    idx = np.arange(n)
    np.random.shuffle(idx)

    df_train = df.iloc[idx[:n_train]].reset_index(drop=True)
    df_val = df.iloc[idx[n_train:n_train + n_val]].reset_index(drop=True)
    df_test = df.iloc[idx[n_train + n_val:]].reset_index(drop=True)
    return df_train, df_val, df_test


features = ['engine_displacement', 'horsepower', 'vehicle_weight', 'model_year']
target = 'fuel_efficiency_mpg'


def prepare_X(df, fill_value):
    # Imputa los nulos con el valor indicado y devuelve la matriz de características
    return df[features].fillna(fill_value).values


def train_linear_regression_reg(X, y, r=0.0):
    # Ecuación normal con Ridge: w = (XᵀX + rI)⁻¹ Xᵀy  (r=0 → sin regularización)
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    XTX = X.T.dot(X) + r * np.eye(X.shape[1])
    w_full = np.linalg.inv(XTX).dot(X.T).dot(y)
    return w_full[0], w_full[1:]


def rmse(y, y_pred):
    return np.sqrt(((y - y_pred) ** 2).mean())


df_train, df_val, df_test = split_dataset(df, seed=42)
y_train = df_train[target].values
y_val = df_val[target].values
len(df_train), len(df_val), len(df_test)
```

## Question 3: Imputación con 0 vs Media (RMSE)

```python
# La media se calcula SOLO con train (evita data leakage) y se aplica a train y a val
mean_hp = df_train['horsepower'].mean()

for name, fill_value in [('0', 0), ('media', mean_hp)]:
    X_train = prepare_X(df_train, fill_value)
    w0, w = train_linear_regression_reg(X_train, y_train, r=0.0)

    X_val = prepare_X(df_val, fill_value)
    y_pred = w0 + X_val.dot(w)
    print(f'Relleno con {name:5s} → RMSE val = {round(rmse(y_val, y_pred), 3)}')
```

## Question 4: Regularización Ridge (mejor parámetro r)

```python
X_train = prepare_X(df_train, 0)
X_val = prepare_X(df_val, 0)

for r in [0, 0.01, 0.1, 1, 5, 10, 100]:
    w0, w = train_linear_regression_reg(X_train, y_train, r=r)
    y_pred = w0 + X_val.dot(w)
    print(f'r = {r:6} → RMSE val = {round(rmse(y_val, y_pred), 4)}')
```

## Question 5: Análisis de estabilidad con diferentes semillas

```python
scores = []

for seed in range(10):
    tr, va, _ = split_dataset(df, seed=seed)
    w0, w = train_linear_regression_reg(prepare_X(tr, 0), tr[target].values, r=0.0)
    y_pred = w0 + prepare_X(va, 0).dot(w)
    score = rmse(va[target].values, y_pred)
    scores.append(score)
    print(f'seed {seed} → RMSE val = {score:.4f}')

print('Desviación típica:', round(np.std(scores), 3))
```

## Question 6: Modelo final (train + val) y evaluación en test

```python
df_train, df_val, df_test = split_dataset(df, seed=9)

# Unimos train + val para entrenar el modelo final con más datos
df_full_train = pd.concat([df_train, df_val]).reset_index(drop=True)
X_full_train = prepare_X(df_full_train, 0)
y_full_train = df_full_train[target].values

w0, w = train_linear_regression_reg(X_full_train, y_full_train, r=0.001)

# Evaluación única en test
y_pred = w0 + prepare_X(df_test, 0).dot(w)
print('RMSE test:', round(rmse(df_test[target].values, y_pred), 3))
```
